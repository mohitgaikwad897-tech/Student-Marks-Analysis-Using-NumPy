```python
#1. Generate marks for 50 students
import numpy as np
marks=np.random.randint(0,101,50)
marks
```

```python
#2. Calculate total, average, highest, lowest.
print("total marks:",marks.sum())
print("average marks:",marks.mean())
print("highest marks:",marks.max())
print("lowest marks:",marks.min())
```

```python
#3. Count above 75, below 35, between 35-75.
above_75=np.sum(marks>75)
below_35=np.sum(marks<35)
between=np.sum((marks>35)&(marks<75))
print("above 75:",above_75)
print("above 35:",below_35)
print("between 35-75:",between)
```

```python
#4. Assign grades.
grade=[]
for i in marks:
    if i >=90:
        grade.append("A+")
    elif i >= 75:
        grade.append("A")
    elif i >= 60:
        grade.append("B")
    elif i >= 35:
        grade.append("C")
    else:
        grade.append("F")
print(grade)
```

```python
#5. Display top 5 marks.
sorted_marks = np.sort(marks)
top5 = sorted_marks[-5:]
print("\nTop 5 Marks:")
print(top5[::-1])
```

```python
#6. Display Bottom 5 Marks
sorted_marks = np.sort(marks)
bottom5 = sorted_marks[:5]
print("\nBottom 5 Marks:")
print(bottom5)
```

```python
#7. Find median and standard deviation
print("median of marks:",np.median(marks))
print("standard deviation:",np.std(marks))
```

```python
#8. Create 5-subject marks matrix and calculate totals
marks = np.random.randint(35, 101, (10, 5))
print("Marks Matrix:")
print(marks)
totals = np.sum(marks, axis=1)
print("\nTotal Marks of Each Student:")
print(totals)
```

```python
#9. Find topper in each subject.
topper = np.max(marks, axis=0)
print("\nHighest Marks in Each Subject:")
print(topper)
```

```python
#10. Generate report card with Total, Average, Percentage and Grade
import numpy as np
marks = np.random.randint(35, 101, (10, 5))
print("Marks Matrix:")
print(marks)
total = np.sum(marks, axis=1)
average = np.mean(marks, axis=1)
percentage = average
print("\nReport Card")
print("-" * 55)
print("Student\tTotal\tAverage\tPercentage\tGrade")
for i in range(len(total)):
    if percentage[i] >= 90:
        grade = "A+"
    elif percentage[i] >= 75:
        grade = "A"
    elif percentage[i] >= 60:
        grade = "B"
    elif percentage[i] >= 35:
        grade = "C"
    else:
        grade = "F"
    print(f"{i+1}\t{total[i]}\t{average[i]:.2f}\t{percentage[i]:.2f}%\t\t{grade}")
```
